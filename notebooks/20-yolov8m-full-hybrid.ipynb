```python
import os, json, glob
from PIL import Image
from tqdm import tqdm

VAL_IMAGES = "/kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/val/images"
VAL_LABELS = "/kaggle/input/datasets/aliabdelmenam/rdd-2022/RDD_SPLIT/val/labels"
COCO_GT_PATH = "/kaggle/working/full_instances_val.json"

categories = [
    {"id": 1, "name": "longitudinal_crack"},
    {"id": 2, "name": "transverse_crack"},
    {"id": 3, "name": "alligator_crack"},
    {"id": 4, "name": "other_corruption"},
    {"id": 5, "name": "pothole"},
]

val_image_files = sorted(glob.glob(f"{VAL_IMAGES}/*.jpg"))
print(f"Converting {len(val_image_files)} validation images to COCO format...")

images, annotations = [], []
annotation_id = 1

for image_path in tqdm(val_image_files):
    image_name = os.path.basename(image_path)
    label_path = os.path.join(VAL_LABELS, image_name.replace(".jpg", ".txt"))
    
    width, height = Image.open(image_path).size
    image_id = image_name.replace(".jpg", "")
    
    images.append({
        "id": image_id,
        "file_name": image_name,
        "width": width,
        "height": height
    })
    
    if os.path.exists(label_path):
        with open(label_path) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) == 5:
                    cls, xc, yc, bw, bh = map(float, parts)
                    x = (xc - bw / 2) * width
                    y = (yc - bh / 2) * height
                    w = bw * width
                    h = bh * height
                    
                    annotations.append({
                        "id": annotation_id,
                        "image_id": image_id,
                        "category_id": int(cls) + 1,
                        "bbox": [float(x), float(y), float(w), float(h)],
                        "area": float(w * h),
                        "iscrowd": 0
                    })
                    annotation_id += 1

coco_gt = {"images": images, "annotations": annotations, "categories": categories}
with open(COCO_GT_PATH, "w") as f:
    json.dump(coco_gt, f)
print(f"Ground Truth COCO saved: {len(annotations)} annotations.")
```

```python
from sahi import AutoDetectionModel
from sahi.predict import get_sliced_prediction

weights_path = "/kaggle/working/repo/runs/detect/thesis_experiments/full_hybrid_50e-2/weights/best.pt"
COCO_PRED_PATH = "/kaggle/working/full_sahi_predictions.json"

detection_model = AutoDetectionModel.from_pretrained(
    model_type='yolov8',
    model_path=weights_path,
    confidence_threshold=0.25,
    device="cuda:0"
)

predictions = []
print("Running SAHI Sliced Inference on 5,758 images...")

for image_path in tqdm(val_image_files):
    image_name = os.path.basename(image_path)
    image_id = image_name.replace(".jpg", "")
    
    result = get_sliced_prediction(
        image_path,
        detection_model,
        slice_height=512,
        slice_width=512,
        overlap_height_ratio=0.2,
        overlap_width_ratio=0.2,
        verbose=0
    )
    
    for pred in result.object_prediction_list:
        x, y, w, h = pred.bbox.to_xywh()
        predictions.append({
            "image_id": image_id,
            "category_id": int(pred.category.id) + 1,
            "bbox": [float(x), float(y), float(w), float(h)],
            "score": float(pred.score.value)
        })

with open(COCO_PRED_PATH, "w") as f:
    json.dump(predictions, f)
print(f"SAHI Predictions saved: {len(predictions)} objects detected.")
```

```python
import os, cv2, random
import pandas as pd
import matplotlib.pyplot as plt
from ultralytics import YOLO
from pycocotools.coco import COCO
from pycocotools.cocoeval import COCOeval
from sahi.predict import get_sliced_prediction
from sahi.utils.cv import visualize_object_predictions

# 1. COCO mAP Evaluation
print("Calculating SAHI mAP Metrics for Full Dataset...")
coco_gt_eval = COCO(COCO_GT_PATH)
coco_dt_eval = coco_gt_eval.loadRes(COCO_PRED_PATH)

coco_eval = COCOeval(coco_gt_eval, coco_dt_eval, iouType="bbox")
coco_eval.evaluate()
coco_eval.accumulate()
coco_eval.summarize()

# 2. Raw Detection Gain Comparison
print("\nCalculating total detection counts (Standard YOLO vs SAHI)...")
model_std = YOLO(weights_path)

total_yolo = 0
for image_path in tqdm(val_image_files):
    total_yolo += len(model_std(image_path, verbose=False)[0].boxes)

total_sahi = len(predictions)
gain_pct = ((total_sahi - total_yolo) / total_yolo) * 100 if total_yolo > 0 else 0

print(f"\n--- DETECTION GAIN SUMMARY ---")
print(f"Standard YOLO Detections: {total_yolo}")
print(f"SAHI Detections:          {total_sahi}")
print(f"Detection Count Gain:     +{gain_pct:.2f}%")

# 3. Generate & Save Top 5 Side-by-Side Visual Comparisons
print("\nGenerating Top 5 visual comparisons...")
top_gains_dir = "/kaggle/working/sahi_full_hybrid_results/top_gains"
os.makedirs(top_gains_dir, exist_ok=True)

sample_files = random.sample(val_image_files, min(300, len(val_image_files)))
sample_gains = []

for image_path in tqdm(sample_files, desc="Finding top gain images"):
    y_cnt = len(model_std(image_path, verbose=False)[0].boxes)
    s_res = get_sliced_prediction(
        image_path, detection_model, slice_height=512, slice_width=512,
        overlap_height_ratio=0.2, overlap_width_ratio=0.2, verbose=0
    )
    s_cnt = len(s_res.object_prediction_list)
    sample_gains.append({
        "path": image_path, "yolo": y_cnt, "sahi": s_cnt,
        "gain": s_cnt - y_cnt, "sahi_res": s_res
    })

top_5_images = sorted(sample_gains, key=lambda x: x["gain"], reverse=True)[:5]

for idx, item in enumerate(top_5_images, start=1):
    img_p = item["path"]
    y_img = model_std(img_p, verbose=False)[0].plot()
    orig = cv2.cvtColor(cv2.imread(img_p), cv2.COLOR_BGR2RGB)
    vis = visualize_object_predictions(
        image=orig, 
        object_prediction_list=item["sahi_res"].object_prediction_list, 
        output_dir=None
    )
    s_img = vis["image"]
    
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    axes[0].imshow(cv2.cvtColor(y_img, cv2.COLOR_BGR2RGB))
    axes[0].set_title(f"Standard YOLO ({item['yolo']} objects)")
    axes[0].axis("off")
    
    axes[1].imshow(s_img)
    axes[1].set_title(f"SAHI Sliced ({item['sahi']} objects)")
    axes[1].axis("off")
    
    plt.tight_layout()
    plt.savefig(f"{top_gains_dir}/top_gain_{idx}.png", dpi=300)
    plt.close()

print(f"Saved Top 5 visual comparison plots to {top_gains_dir}")

# 4. Save Summary & Create Zip
summary = {
    "Standard_YOLO_Detections": total_yolo,
    "SAHI_Detections": total_sahi,
    "Detection_Gain_Pct": gain_pct,
    "Precision_mAP50": coco_eval.stats[1],
    "mAP50-95": coco_eval.stats[0],
    "Recall_AR100": coco_eval.stats[8],
}
df = pd.DataFrame([summary])
df.to_csv("/kaggle/working/sahi_full_hybrid_results/sahi_full_hybrid_metrics.csv", index=False)
display(df)

```
